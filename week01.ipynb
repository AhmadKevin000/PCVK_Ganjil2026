```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
imageurls = [
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/starry_night.jpg"
]

for url in imageurls:
    image = io.imread(url)                          #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
    final_frame = cv.hconcat((image, image_2))      #concatenate image
    cv2_imshow(final_frame)                         #show image
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar ", tinggi," x ", lebar)
cv2_imshow(image_2)
```

```python
image_3 = image_2.copy()
row = int(image_3.shape[0] / 2)

for x in range(image_3.shape[1]):
    image_3[row, x] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
img_vertikal = image_2.copy()
img_diagonal = image_2.copy()

tinggi = image_2.shape[0]
lebar = image_2.shape[1]
col_tengah = lebar // 2

# Membuat garis vertikal
for y in range(tinggi):
    img_vertikal[y, col_tengah] = [0, 255, 0]

# Membuat garis menyilang diagonal
for y in range(tinggi):
    x1 = int(y * (lebar / tinggi))
    x2 = lebar - 1 - x1

    if x1 < lebar:
        img_diagonal[y, x1] = [0, 255, 0]
    if x2 >= 0:
        img_diagonal[y, x2] = [0, 255, 0]

final_tugas2 = cv.hconcat((img_vertikal, img_diagonal))
print("Hasil Tugas 2: Garis Vertikal & Menyilang Diagonal")
cv2_imshow(final_tugas2)
```

```python
img_tugas3 = image_2.copy()
tinggi = img_tugas3.shape[0]
row_tengah = tinggi // 2

for x in range(50, 250):
    img_tugas3[row_tengah, x] = [0, 0, 255]

print("Hasil Tugas 3: Garis Horizontal Merah")
cv2_imshow(img_tugas3)
```

```python
img_tugas4 = image_2.copy()

# Membuat kotak putih di koordinat tertentu
img_tugas4[100:200, 150:250] = [255, 255, 255]

print("Hasil Tugas 4: Kotak Putih")
cv2_imshow(img_tugas4)
```
